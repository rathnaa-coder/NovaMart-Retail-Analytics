# Novamart Retail Sales - Visualization

## Project Overview

This notebook visualizes key business insights from the NovaMart Retail Sales dataset.

The purpose of these visualizations is to identify sales trends, customer preferences, and business performance before building an interactive Power BI dashboard.

### Visualizations Included

- Sales by Product Category
- Sales by City
- Payment Method Distribution
- Monthly Sales Trend

## Data Source

The dataset used in this notebook was synthetically generated using Python as part of the NovaMart Retail Analytics project.

Dataset Size:

- Rows : 10,000
- Columns : 17

## Tools Used

- Python
- Pandas
- Matplotlib
- Jupyter Notebook

# Step 1 - Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

# Step 2 - Load the dataset

```python
df = pd.read_csv("novamart_sales_dataset.csv")
df["order_date"] = pd.to_datetime(df["order_date"])
```

# Step 3 - Sales by Product Category

### Business Question

Which product category contributes the highest sales revenue?

```python
category_sales = (
    df.groupby("category")["sales"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8,5))
plt.bar(category_sales.index, category_sales.values)

plt.title("Total Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales (₹)")

plt.xticks(rotation=20)

plt.tight_layout()
plt.savefig("sales_by_category.png", dpi=300)
plt.show()
```

### Observation

Electronics generated the highest sales among all product categories.

This indicates strong customer demand for electronic products.

# Step 4 - Sales by City

### Business Question

Which city generated the highest revenue?

```python
city_sales = (
    df.groupby("city")["sales"]
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))
plt.bar(city_sales.index, city_sales.values)

plt.title("Total Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales (₹)")

plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("Total_sales_by_city.png", dpi=300)
plt.show()
```

### Observation

Coimbatore generated the highest sales revenue among all cities.

This indicates that Coimbatore is currently the strongest-performing market.

# Step 5 - Payment Method Distribution

### Business Question

Which payment method is used most frequently by customers?

```python
payment_counts = df["payment_method"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(
    payment_counts.values,
    labels=payment_counts.index,
    autopct="%1.1f%%"
)

plt.title("Payment Method Distribution")

plt.tight_layout()
plt.savefig("Payment_Method_Distribution.png", dpi=300)
plt.show()
```

### Observation

UPI is the most preferred payment method, indicating a strong customer preference for digital payments.

# Step 6 - Monthly Sales Trend

### Business Question

How do sales vary across different months?

```python
monthly_sales = (
    df.groupby(df["order_date"].dt.month_name())["sales"]
      .sum()
      .reindex([
          "January","February","March","April",
          "May","June","July","August",
          "September","October","November","December"
      ])
)

plt.figure(figsize=(10,5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales (₹)")

plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("Monthly_sales_trend.png", dpi=300)
plt.show()
```

### Observation

The highest sales were recorded in November.

The lowest sales were recorded in February.

This helps us to identify seasonal demand patterns.

# Conclusion

The visualizations highlight key trends in product performance, customer payment preferences, city-wise sales, and monthly revenue.

These insights provide a strong foundation for business decision-making and will be used to build an interactive Power BI dashboard for deeper analysis.
