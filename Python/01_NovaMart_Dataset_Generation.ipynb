# Step 4 - Generate Customer Master

This section generates 6000 synthetic customers.

The customer master is later used to simulate repeat purchases.

```python
customers = []

for i in range(1,6001):

    city = random.choice(list(cities.keys()))
    state = cities[city]

    customers.append({
        "customer_id": f"CUST{i:04}",
        "customer_name": fake.name(),
        "city": city,
        "state": state
    })

customer_master = pd.DataFrame(customers)

customer_master.head()
```

# Step 5 - Generate Transaction Function

This function generates one complete retail transaction.

Each transaction contains:

- Customer
- Product
- Quantity
- Sales
- Profit
- Payment Method
- Region

```python
def generate_transaction(order_number):

    # Select a random product category
    category = random.choice(list(product_master.keys()))
    product = random.choice(list(product_master[category].keys()))
    selling_price, cost_price = product_master[category][product]

    # Order
    order_id = f"ORD{order_number:06d}"

    customer = customer_master.sample(1).iloc[0]

    customer_id = customer["customer_id"]

    customer_name = customer["customer_name"]

    city = customer["city"]

    state = customer["state"]
    
    region_mapping = {
    "Tamil Nadu": "South",
    "Karnataka": "South",
    "Kerala": "South",
    "Telangana": "South",
    "Andhra Pradesh": "South",

    "Maharashtra": "West",
    "Gujarat": "West",
    "Rajasthan": "West",
    "Goa": "West",

    "Delhi": "North",
    "Punjab": "North",
    "Haryana": "North",
    "Uttar Pradesh": "North",

    "West Bengal": "East",
    "Odisha": "East",
    "Bihar": "East",
    "Jharkhand": "East",
    "Assam": "East"
}

    region = region_mapping[state]

    # Date
    start_date = datetime(2025,1,1)
    end_date = datetime(2025,12,31)

    random_days = random.randint(0,(end_date-start_date).days)
    order_date = start_date + timedelta(days=random_days)

    # Sales Details
    quantity = random.randint(1,5)

    discount = random.choice([0,5,10,15,20])

    sales = selling_price * quantity

    discount_amount = sales * discount / 100

    final_sales = sales - discount_amount

    cost = cost_price * quantity

    # Calculate profit
    profit = final_sales - cost

    # Payment & Salesperson
    payment_method = random.choices(
        payment_methods,
        weights=[45,20,15,10,10],
        k=1
    )[0]

    salesperson = random.choice(salespersons)

    # Return one transaction
    return {
        "order_id": order_id,
        "order_date": order_date.strftime("%Y-%m-%d"),
        "customer_id": customer_id,
        "customer_name": customer_name,
        "city": city,
        "state": state,
        "region": region,
        "category": category,
        "product_name": product,
        "quantity": quantity,
        "unit_price": selling_price,
        "discount_percent": discount,
        "sales": round(final_sales,2),
        "cost": cost,
        "profit": round(profit,2),
        "payment_method": payment_method,
        "salesperson": salesperson
    }
```
