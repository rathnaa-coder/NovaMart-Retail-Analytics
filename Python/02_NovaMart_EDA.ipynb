# Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

# Step 1 - Load Dataset

```python
df = pd.read_csv("novamart_sales_dataset.csv")

df.head()
```

```python
df.shape

print(f"Rows: {df.shape[0]}")
print(f"Columns: {df.shape[1]}")
```

```python
df.describe()
```

# Step 3 - Business KPIs

Calculation of
  1. Total Sales
  2. Total Profit
  3. Average Order Value

```python
total_sales = df["sales"].sum()
total_profit = df["profit"].sum()
average_order = df["sales"].mean()

print(f"💰 Total Sales : ₹{total_sales:,.2f}")
print(f"📈 Total Profit : ₹{total_profit:,.2f}")
print(f"🛒 Average Order Value : ₹{average_order:,.2f}")
```

### Observation

- The dataset generated total sales of ₹ 489,890,040
- The business achieved a total profit of ₹ 45,750,800
- The average order value indicates the typical customer spending per transaction.

# Step 4 - Product Category Analysis

Which category earns the most?

```python
df.groupby("category")["sales"].sum().sort_values(ascending=False)
```

### Insight

Electronics generated the highest sales revenue.

This suggests customers spend more on high-value electronic products than other categories.

# Step 5 - Payment Method Analysis

Which payment method is most popular?

```python
df["payment_method"].value_counts()
```

### Insight

UPI was the most preferred payment method, indicating customers prefer digital transactions.

# Step 6 - City-wise Sales Analysis

Which city generated the highest sales?

```python
df.groupby("city")["sales"].sum().sort_values(ascending=False)
```

### Insight

Coimbatore generated the highest revenue, making it the strongest performing market.

# Step 7 - Salesperson Performance

Who generated the most revenue?

```python
df.groupby("salesperson")["sales"].sum().sort_values(ascending=False)
```

### Insight

The highest-performing salesperson Priyanka generated the maximum sales revenue and can be studied to understand best sales practices.

# Step 8 - Monthly Sales Analysis

```python
df["order_date"] = pd.to_datetime(df["order_date"])

monthly_sales = (
    df.groupby(df["order_date"].dt.month_name())["sales"]
      .sum()
      .reindex([
          "January","February","March","April",
          "May","June","July","August",
          "September","October","November","December"
      ])
)

monthly_sales
```

### Insight

Highest Sales Month : November

Lowest Sales Month : February

# Step 9 - Top Product Analysis

```python
df.groupby("product_name")["sales"].sum().sort_values(ascending=False).head(10)
```

### Insight

The products listed above contributed the highest sales revenue and represent the company's strongest-performing items.

# Step 10 - Customers Analysis

```python
df.groupby("customer_name")["sales"].sum().sort_values(ascending=False).head(10)
```

### Insight

These customers generated the highest sales value and could be considered valuable repeat customers.

# Step 11 - Profitability Analysis

```python
df.groupby("category")["profit"].mean().sort_values(ascending=False)
```

### Insight

This analysis identifies which product category provides the highest average profitability rather than just the highest sales.

# Step 12 - Average Order Value Analysis

```python
df.groupby("payment_method")["sales"].mean().sort_values(ascending=False)
```

### Insight

Customers using Debit card tend to place higher-value orders on average.
